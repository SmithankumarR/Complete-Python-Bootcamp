```python
#Tokenization Example
```

```python
corpus="""This is a simple example of a corpus for tokenization. 
It contains multiple sentences, punctuation, and various words. 
The goal is to demonstrate how tokenization !
works in natural language processing."""
```

```python
corpus = corpus.replace("\n", " ")
print(corpus)
```

```python
#tokenization
## sentence --> paragraphs
import nltk
#nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(corpus)
print(sentences)
```

```python
type(sentences)

for sentence in sentences:
    print(sentence)
```

```python
#Tokenization
## paragraph --- words
## sentence --- words

from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)

type(words)

for sentence in sentences:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize #tokeinzes including punctation marks

wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer #fullstop won't be considered in middle only at the end 

tokenizer = TreebankWordTokenizer()

tokenizer.tokenize(corpus)
```
